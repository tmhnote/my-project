```python
import numpy as np
import matplotlib.pyplot as plt
# 图片内嵌显示
%matplotlib inline
```

```python
x = np.arange(0, 6, 0.1)
y = np.sin(x)
```

```python
# 生成数据
y1 = np.sin(x)
y2 = np.cos(x)
```

```python
# 绘图
plt.plot(x, y1, label='sin(x)')
plt.plot(x, y2, linestyle = '--', label= 'cos(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.title('sin & cos')
plt.legend()
plt.show()
```
