# 感知机

## 简单实现

```python
# 与门
def AND(x1, x2):
    w1, w2, theta = 0.5, 0.5, 0.7
    tmp = x1*w1 + x2*w2
    if tmp <= theta:
        return 0
    elif tmp > theta:
        return 1
```

```python
AND(0, 0)


```

```python
AND(0, 1)
```

```python
AND(1, 0)
```

```python
AND(1, 1)
```

### sigmoid函数

```python
import numpy as np

def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

### 阶跃函数

```python
# 简单实现
import numpy as np

def step_function(x):
    if x > 0:
        return 1
    else:
        return 0
```

上面参数x 只能接受实数，不允许参数为数组

进行修改为支持数组的实现

```python
def step_function(x):
    y = x > 0:
    return y.astype(np.int)
```

```python
import numpy as np

x = np.array([-1.0, 1.0, 2.0])
x
```

```python
y = x > 0
y
```

从右往左来，`x > 0` 会对数组 `x` 里的**每一个元素**，逐个执行「是否大于 0」的判断，最终返回一个和 `x` 形状完全相同的**布尔数组**

`y = x > 0`：把生成的布尔数组赋值给变量 `y`。

输出 `y` 的结果：`array([False,  True,  True])`。

这类布尔数组最常用的场景是**布尔索引**，比如执行 `x[y]`，就可以直接取出 `x` 中所有大于 0 的元素，得到 `array([1., 2.])`。
